### Problem 1: Whitespace tokenize and count

```python
def tok(text): return text.lower().split()
s = "Attention is all you need"
print("tokens:", tok(s))
print("count:", len(tok(s)))
```

### Problem 2: Two tokenizers, one word

```python
word = "playing"
whitespace = [word]
suffix = [word[:-3], "##ing"] if word.endswith("ing") else [word]
print("whitespace:", whitespace)
print("suffix:", suffix)
```

### Problem 3: Encode to ids and decode back

```python
vocab = {"i": 0, "love": 1, "nlp": 2}
inv = {v: k for k, v in vocab.items()}
tokens = ["i", "love", "nlp", "love"]
ids = [vocab[t] for t in tokens]
print("ids:", ids)
print("decoded:", [inv[i] for i in ids])
```

### Problem 4: Token count is not word count

```python
text = "cats running"           # 2 words
tokens = ["cat", "##s", "runn", "##ing"]   # a subword tokenizer's output
print("words:", len(text.split()))
print("tokens:", len(tokens))
```

### Problem 5: A bag of tokens loses order

```python
a = "dog bites man"
b = "man bites dog"
print("same bag:", sorted(a.split()) == sorted(b.split()))
print("same sentence:", a == b)
```

### Problem 6: Why order matters

```python
print("a bag of independent tokens keeps which words appear but not their order, "
      "so it cannot tell 'dog bites man' from 'man bites dog'; attention restores order and context")
```
