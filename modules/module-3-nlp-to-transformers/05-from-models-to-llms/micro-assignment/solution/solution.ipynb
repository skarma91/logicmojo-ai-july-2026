### Problem 1: Softmax with temperature

```python
import numpy as np
def softmax_T(logits, T):
    z = np.array(logits) / T
    e = np.exp(z - z.max()); return e / e.sum()
print("probs:", np.round(softmax_T([2., 1., 0.], 1.0), 3))
```

### Problem 2: Lower temperature is sharper

```python
print("T=0.5 max:", round(float(softmax_T([2., 1., 0.], 0.5).max()), 2))
print("T=2   max:", round(float(softmax_T([2., 1., 0.], 2.0).max()), 2))
```

### Problem 3: Nucleus size for top-p

```python
probs = [0.5, 0.25, 0.15, 0.1]      # already sorted, descending
total, size = 0.0, 0
for pr in probs:
    size += 1; total += pr
    if total >= 0.9:
        break
print("nucleus size (p=0.9):", size)
```

### Problem 4: Temperature 0 takes the top token

```python
logits = [2., 1., 3.]
print("chosen index:", int(np.argmax(logits)))
```
