### Problem 1: Feed-forward parameter count

```python
d = 8
# Linear(d, 4d) then Linear(4d, d), each with weights and biases
params = (d*(4*d) + 4*d) + ((4*d)*d + d)
print("feed-forward params:", params)
```

### Problem 2: Head dimension

```python
d_model, n_heads = 12, 4
print("per-head dimension:", d_model // n_heads)
```

### Problem 3: A residual connection

```python
import numpy as np
x = np.array([1., 2., 3.])
layer_out = np.array([0.1, 0.1, 0.1])
print("residual:", (x + layer_out).tolist())
```

### Problem 4: Concatenate two heads

```python
h1 = np.array([1., 2.]); h2 = np.array([3., 4.])
print("combined:", np.concatenate([h1, h2]).tolist())
```

### Problem 5: LayerNorm centering

```python
v = np.array([2., 4., 6.])
print("mean:", float(v.mean()))
print("centered:", (v - v.mean()).tolist())
```
