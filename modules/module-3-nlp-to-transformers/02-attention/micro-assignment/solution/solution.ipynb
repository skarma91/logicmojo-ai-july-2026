# Micro-assignment 3.2 solution: Attention

### Problem 1: A similarity score (dot product)

```python
import numpy as np
q = np.array([2., 0., 1.])
k = np.array([1., 3., 1.])
print("score:", float(q @ k))
```

### Problem 2: Softmax over scores

```python
s = np.array([4., 1., 0.])
w = np.exp(s - s.max()); w = w / w.sum()
print("weights:", np.round(w, 3))
print("sum:", round(float(w.sum()), 3))
```

### Problem 3: Scale by sqrt(d_k)

```python
scores = np.array([8., 4., 2.])
d_k = 4
print("scaled:", (scores / np.sqrt(d_k)).tolist())
```

### Problem 4: The context vector (weighted blend)

```python
weights = np.array([0.9, 0.1])
values = np.array([[1., 0.], [0., 1.]])
print("context:", (weights @ values).tolist())
```

### Problem 5: Where does this token attend?

```python
row = np.array([0.1, 0.7, 0.2])
print("most-attended index:", int(np.argmax(row)))
```

### Problem 6: Why scale the scores?

```python
print("dividing by sqrt(d_k) keeps dot products from growing large with vector length, "
      "so softmax stays smooth and training is stable")
```

### Problem 7: Causal mask (no peeking ahead)

A query at position 1 may attend to positions 0 and 1 only. Set positions 2 and 3 to `-inf`, softmax, and print the weights (three decimals).

```python
import numpy as np
scores = np.array([2.0, 1.0, 3.0, 0.5])
scores[2:] = -np.inf                       # block future positions
e = np.exp(scores - scores[np.isfinite(scores)].max())
weights = e / e.sum()
print("weights:", np.round(weights, 3))
```
