### Problem 1: Tensors and broadcasting

```python
import torch
t    = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
bias = torch.tensor([10., 20., 30.])
print("shape:", tuple(t.shape))
print("dtype:", t.dtype)
print("broadcast:", (t + bias).tolist())
```

### Problem 2: Autograd matches the by-hand gradient

```python
x = torch.tensor([1., 2., 3.], requires_grad=True)
y = (x ** 2).sum()
y.backward()
print("x.grad:", x.grad)      # 2*x = [2., 4., 6.]
```

### Problem 3: Count the parameters of an nn.Module

```python
import torch.nn as nn
net = nn.Sequential(nn.Linear(64, 16), nn.ReLU(), nn.Linear(16, 10))
print("total parameters:", sum(p.numel() for p in net.parameters()))
```

### Problem 4: The five-line loop lowers the loss

```python
x = torch.tensor([1., 2., 3., 4.])
y = torch.tensor([2., 4., 6., 8.])
w = torch.tensor([0.0], requires_grad=True)
opt = torch.optim.SGD([w], lr=0.02)
loss_fn = nn.MSELoss()

first = None
for step in range(80):
    pred = w * x
    loss = loss_fn(pred, y)
    if step == 0:
        first = round(loss.item(), 2)
    opt.zero_grad()
    loss.backward()
    opt.step()
last = round(loss_fn(w * x, y).item(), 2)
print("first loss:", first)
print("last loss:", last)
print("decreased:", last < first)
```

### Problem 5: The zero_grad trap

```python
w = torch.tensor([1.0], requires_grad=True)
without = []
for _ in range(3):
    (w ** 2).sum().backward()
    without.append(float(w.grad))

w = torch.tensor([1.0], requires_grad=True)
withz = []
for _ in range(3):
    if w.grad is not None:
        w.grad.zero_()
    (w ** 2).sum().backward()
    withz.append(float(w.grad))

print("without zero_grad:", without)
print("with zero_grad:", withz)
```
