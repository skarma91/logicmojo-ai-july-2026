### Problem 1: A neuron forward pass

```python
import numpy as np
w = np.array([0.5, -0.2, 0.1]); x = np.array([2, 1, 4]); b = 0.3
z = w @ x + b
print("z:", round(float(z), 2))
print("ReLU(z):", round(float(max(0.0, z)), 2))
```

### Problem 2: Apply ReLU, sigmoid, and tanh

```python
v = np.array([-2.0, 0.0, 3.0])
print("ReLU:", np.round(np.maximum(0, v), 2))
print("sigmoid:", np.round(1 / (1 + np.exp(-v)), 2))
print("tanh:", np.round(np.tanh(v), 2))
```

### Problem 3: A layer of two neurons

```python
W = np.array([[0.5, -1.0], [1.0, 0.5]])
b = np.array([0.0, 1.0])
x = np.array([2.0, 3.0])
outputs = np.maximum(0, W @ x + b)
print("outputs:", np.round(outputs, 2))
```

### Problem 4: Two linear layers collapse to one

```python
W1 = np.array([[1.0, 2.0], [0.0, 1.0]]); b1 = np.array([0.0, 0.0])
W2 = np.array([[2.0, -1.0]]);            b2 = np.array([0.0])
x = np.array([1.0, 1.0])
two = W2 @ (W1 @ x + b1) + b2
one = (W2 @ W1) @ x + (W2 @ b1 + b2)
print("two layers:", np.round(two, 3))
print("one layer:", np.round(one, 3))
print("identical:", np.allclose(two, one))
```

### Problem 5: Softmax

```python
logits = np.array([1.0, 2.0, 3.0])
e = np.exp(logits - logits.max())
probs = e / e.sum()
print("probs:", np.round(probs, 3))
print("sum:", round(float(probs.sum()), 3))
```

### Problem 6: Cross-entropy against a one-hot target

```python
y = np.array([0, 0, 1])            # one-hot target, true class is index 2
probs = np.array([0.1, 0.2, 0.7])
ce = -np.sum(y * np.log(probs))
true_class = int(np.argmax(y))
print("cross-entropy:", round(float(ce), 2))
print("same as -log(p_true):", round(float(-np.log(probs[true_class])), 2))
print("note: cross-entropy grows toward infinity as the probability on the true class nears 0, so a confident wrong answer is punished hardest")
```
