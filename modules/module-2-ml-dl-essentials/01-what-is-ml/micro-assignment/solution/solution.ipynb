### Problem 2: Three-way split and model selection

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X, y = load_wine(return_X_y=True)
X_tv, X_te, y_tv, y_te = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)
X_tr, X_val, y_tr, y_val = train_test_split(X_tv, y_tv, test_size=0.25, random_state=0, stratify=y_tv)

scaler = StandardScaler().fit(X_tr)                      # fit on train only
X_tr_s, X_val_s, X_te_s = scaler.transform(X_tr), scaler.transform(X_val), scaler.transform(X_te)

logreg = LogisticRegression(max_iter=5000).fit(X_tr_s, y_tr)
tree = DecisionTreeClassifier(max_depth=2, random_state=0).fit(X_tr_s, y_tr)
v_log = round(accuracy_score(y_val, logreg.predict(X_val_s)), 2)
v_tree = round(accuracy_score(y_val, tree.predict(X_val_s)), 2)

chosen_name, chosen = ("logreg", logreg) if v_log >= v_tree else ("tree", tree)
print("sizes train/val/test:", X_tr.shape[0], X_val.shape[0], X_te.shape[0])
print("validation logreg:", v_log)
print("validation tree:", v_tree)
print("chosen:", chosen_name)
print("test accuracy:", round(accuracy_score(y_te, chosen.predict(X_te_s)), 2))
```

### Problem 6: Precision and recall on imbalanced labels

```python
labels = [0] * 18 + [1, 1]
preds  = [0] * 20

TP = sum(1 for l, p in zip(labels, preds) if l == 1 and p == 1)
FP = sum(1 for l, p in zip(labels, preds) if l == 0 and p == 1)
FN = sum(1 for l, p in zip(labels, preds) if l == 1 and p == 0)

accuracy = sum(1 for l, p in zip(labels, preds) if l == p) / len(labels)
precision = TP / (TP + FP) if (TP + FP) else 0.0
recall = TP / (TP + FN) if (TP + FN) else 0.0

print("accuracy:", round(accuracy, 2))
print("precision:", round(precision, 2))
print("recall:", round(recall, 2))
print("note: accuracy is high, but recall 0.0 means it caught none of the rare positives")
```
