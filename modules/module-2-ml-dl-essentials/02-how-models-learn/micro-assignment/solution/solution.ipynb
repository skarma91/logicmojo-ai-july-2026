### Problem 1: Mean squared error by hand

```python
y_hat = [4.0, 1.0, 3.0, 6.0]
y     = [3.5, 2.0, 3.0, 5.0]
mse = sum((a - b)**2 for a, b in zip(y_hat, y)) / len(y_hat)
print(f"MSE: {mse:.2f}")
```

### Problem 2: One descent step

```python
theta, grad, eta = 0.8, 2.0, 0.1
theta_new = theta - eta*grad
print("theta_new:", round(theta_new, 2))
```

### Problem 3: A five-step descent

```python
x = 0.0
lr = 0.3
xs = [round(x, 3)]
for _ in range(5):
    x = x - lr * 2*(x - 3)      # slope of (x-3)**2 is 2*(x-3)
    xs.append(round(x, 3))
print("x each step:", xs)
```

### Problem 4: Diagnose three learning rates

```python
def final_and_start(lr, x0=1.0, steps=15):
    x = x0
    start = x*x
    for _ in range(steps):
        x = x - lr * 2*x        # slope of x**2 is 2*x
    return start, x*x

labels = []
for lr in [0.05, 0.4, 1.1]:
    start, final = final_and_start(lr)
    if final > start:
        labels.append("diverges")
    elif final < 0.01:
        labels.append("converges")
    else:
        labels.append("crawls")
print(labels)
```

### Problem 6: Cross-entropy on a confident-wrong guess

```python
import math
def bce(p):                     # true label is 1, so loss is -log(p)
    return -math.log(p)
print(f"cross-entropy at p=0.9: {bce(0.9):.2f}")
print(f"cross-entropy at p=0.1: {bce(0.1):.2f}")
print("note: the confident-wrong guess costs far more, which is why classification uses cross-entropy")
```
